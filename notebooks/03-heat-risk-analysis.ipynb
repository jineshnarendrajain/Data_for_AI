# Land Surface Temperature Assessment

Land Surface Temperature (LST) was derived from Landsat 8 thermal imagery.

LST provides a measure of surface heating conditions and is widely used in urban heat island studies.

Higher values indicate greater thermal exposure.

```python
lst = profile["lst_statistics"]

lst_table = pd.DataFrame({
    "Statistic": ["Minimum", "Mean", "Maximum"],
    "LST (°C)": [
        round(lst["LST_min"], 2),
        round(lst["LST_mean"], 2),
        round(lst["LST_max"], 2)
    ]
})

lst_table
```

# Vegetation Density Assessment

Vegetation density was assessed using the Normalized Difference Vegetation Index (NDVI).

NDVI is a widely used indicator of vegetation coverage and environmental quality.

Vegetation can help reduce urban heat through shading and evapotranspiration.

```python
ndvi = profile["ndvi_statistics"]

ndvi_table = pd.DataFrame({
    "Statistic": ["Minimum", "Mean", "Maximum"],
    "NDVI": [
        round(ndvi["NDVI_min"], 2),
        round(ndvi["NDVI_mean"], 2),
        round(ndvi["NDVI_max"], 2)
    ]
})

ndvi_table
```

# Heat Risk Framework

The project combines temperature and vegetation information into a single interpretable framework.

Areas with:

- Higher temperatures
- Lower vegetation density

are expected to exhibit greater relative heat exposure.

The framework uses normalized environmental indicators to calculate a heat risk score.

```python
print("Heat Risk Score")
print()
print("Risk = Normalized LST - Normalized NDVI")
```

# Heat Risk Logic Demonstration

The following examples illustrate how the framework behaves under different environmental conditions.

Higher temperatures increase risk.

Greater vegetation reduces risk.

```python
examples = pd.DataFrame({
    "Scenario": [
        "Cool & Green",
        "Moderate Conditions",
        "Hot & Sparse Vegetation"
    ],
    "Normalized_LST": [
        0.20,
        0.60,
        0.95
    ],
    "Normalized_NDVI": [
        0.85,
        0.50,
        0.10
    ]
})

examples["Heat_Risk_Score"] = (
    examples["Normalized_LST"]
    - examples["Normalized_NDVI"]
)

examples
```
